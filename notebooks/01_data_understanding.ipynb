```python
import pandas as pd

mat_df = pd.read_csv("../data/raw/student-mat.csv", sep=";")
por_df = pd.read_csv("../data/raw/student-por.csv", sep=";")

print("Math Dataset Shape:", mat_df.shape)
print("Portuguese Dataset Shape:", por_df.shape)

mat_df.head()
```

```python
print("Math Dataset Missing Values")
print(mat_df.isnull().sum())

print("\n" + "="*50 + "\n")

print("Portuguese Dataset Missing Values")
print(por_df.isnull().sum())
```

```python
print("Math Dataset Duplicate Rows:")
print(mat_df.duplicated().sum())

print("\nPortuguese Dataset Duplicate Rows:")
print(por_df.duplicated().sum())
```

```python
# Create copies

mat_clean = mat_df.copy()
por_clean = por_df.copy()

print("Math Shape:", mat_clean.shape)
print("Portuguese Shape:", por_clean.shape)
```

```python
mat_clean.to_csv(
    "../data/processed/student_mat_clean.csv",
    index=False
)

por_clean.to_csv(
    "../data/processed/student_por_clean.csv",
    index=False
)

print("Clean datasets saved successfully.")
```
