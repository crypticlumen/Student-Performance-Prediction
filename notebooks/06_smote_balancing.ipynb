# Phase 11: SMOTE Class Balancing

## Objective
Balance the Pass and At Risk classes in the training data using SMOTE.

## Research safeguards
- Apply SMOTE only to training data.
- Never resample the test set.
- Preserve the original test features and labels.
- Save balanced training data separately.
- Compare class distributions before and after SMOTE.

```python
from pathlib import Path

import pandas as pd
from imblearn.over_sampling import SMOTE

print("Libraries imported successfully.")
```

```python
PROJECT_ROOT = Path.cwd()

# If the notebook's working directory is notebooks/,
# move to the project root.
if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

SPLIT_DIR = PROJECT_ROOT / "data" / "processed" / "splits"
SCALED_DIR = PROJECT_ROOT / "data" / "processed" / "scaled"
OUTPUT_DIR = PROJECT_ROOT / "data" / "processed" / "smote"

required_files = [
    SPLIT_DIR / "mat_y_train.csv",
    SPLIT_DIR / "por_y_train.csv",
    SCALED_DIR / "mat_X_train_scaled.csv",
    SCALED_DIR / "por_X_train_scaled.csv",
]

for file_path in required_files:
    print(f"{file_path.name}: {'FOUND' if file_path.exists() else 'MISSING'}")

assert all(path.exists() for path in required_files), (
    "A required file is missing. Check the paths before proceeding."
)

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

print("\nAll required files found.")
print("Output directory:", OUTPUT_DIR)
```

```python
mat_X_train = pd.read_csv(SCALED_DIR / "mat_X_train_scaled.csv")
mat_y_train = pd.read_csv(SPLIT_DIR / "mat_y_train.csv").iloc[:, 0]

por_X_train = pd.read_csv(SCALED_DIR / "por_X_train_scaled.csv")
por_y_train = pd.read_csv(SPLIT_DIR / "por_y_train.csv").iloc[:, 0]

print("Mathematics training features:", mat_X_train.shape)
print("Mathematics training labels:", mat_y_train.shape)

print("Portuguese training features:", por_X_train.shape)
print("Portuguese training labels:", por_y_train.shape)

assert len(mat_X_train) == len(mat_y_train)
assert len(por_X_train) == len(por_y_train)

assert mat_y_train.notna().all()
assert por_y_train.notna().all()

print("\nTraining data loaded and verified.")
```

```python
print("MATHEMATICS — before SMOTE")
print(mat_y_train.value_counts().sort_index())

print("\nPORTUGUESE — before SMOTE")
print(por_y_train.value_counts().sort_index())
```

```python
# Use the same random seed for reproducibility.
# k_neighbors is set to 5, the standard SMOTE default.

smote_mat = SMOTE(random_state=42, k_neighbors=5)
smote_por = SMOTE(random_state=42, k_neighbors=5)

mat_X_balanced_array, mat_y_balanced_array = smote_mat.fit_resample(
    mat_X_train,
    mat_y_train
)

por_X_balanced_array, por_y_balanced_array = smote_por.fit_resample(
    por_X_train,
    por_y_train
)

# Convert back to DataFrames to preserve feature names.
mat_X_balanced = pd.DataFrame(
    mat_X_balanced_array,
    columns=mat_X_train.columns
)
mat_y_balanced = pd.Series(mat_y_balanced_array, name="performance")

por_X_balanced = pd.DataFrame(
    por_X_balanced_array,
    columns=por_X_train.columns
)
por_y_balanced = pd.Series(por_y_balanced_array, name="performance")

print("SMOTE applied successfully.")
print("Mathematics balanced shape:", mat_X_balanced.shape)
print("Portuguese balanced shape:", por_X_balanced.shape)
```

```python
from pathlib import Path

import pandas as pd
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from imblearn.over_sampling import SMOTE

PROJECT_ROOT = (
    Path.cwd().parent
    if Path.cwd().name == "notebooks"
    else Path.cwd()
)

SPLIT_DIR = PROJECT_ROOT / "data" / "processed" / "splits"
PROCESSED_DIR = PROJECT_ROOT / "data" / "processed"

print("Project root:", PROJECT_ROOT)
print("Split directory exists:", SPLIT_DIR.exists())
```

```python
mat_X_train = pd.read_csv(SPLIT_DIR / "mat_X_train.csv")
mat_X_test = pd.read_csv(SPLIT_DIR / "mat_X_test.csv")
mat_y_train = pd.read_csv(SPLIT_DIR / "mat_y_train.csv").iloc[:, 0]
mat_y_test = pd.read_csv(SPLIT_DIR / "mat_y_test.csv").iloc[:, 0]

por_X_train = pd.read_csv(SPLIT_DIR / "por_X_train.csv")
por_X_test = pd.read_csv(SPLIT_DIR / "por_X_test.csv")
por_y_train = pd.read_csv(SPLIT_DIR / "por_y_train.csv").iloc[:, 0]
por_y_test = pd.read_csv(SPLIT_DIR / "por_y_test.csv").iloc[:, 0]

print("Mathematics:", mat_X_train.shape, mat_X_test.shape)
print("Portuguese:", por_X_train.shape, por_X_test.shape)

print("\nMathematics training labels:")
print(mat_y_train.value_counts().sort_index())

print("\nPortuguese training labels:")
print(por_y_train.value_counts().sort_index())
```

```python
def preprocess_features(X_train, X_test):
    categorical_cols = X_train.select_dtypes(
        include=["object", "category", "bool"]
    ).columns.tolist()

    numeric_cols = X_train.select_dtypes(
        include=["number"]
    ).columns.tolist()

    preprocessor = ColumnTransformer(
        transformers=[
            (
                "categorical",
                OneHotEncoder(handle_unknown="ignore"),
                categorical_cols,
            ),
            (
                "numeric",
                StandardScaler(),
                numeric_cols,
            ),
        ],
        remainder="drop",
    )

    X_train_processed = preprocessor.fit_transform(X_train)
    X_test_processed = preprocessor.transform(X_test)

    feature_names = preprocessor.get_feature_names_out()

    X_train_processed = pd.DataFrame(
        X_train_processed,
        columns=feature_names,
        index=X_train.index,
    )

    X_test_processed = pd.DataFrame(
        X_test_processed,
        columns=feature_names,
        index=X_test.index,
    )

    return X_train_processed, X_test_processed, preprocessor


mat_X_train_processed, mat_X_test_processed, mat_preprocessor = (
    preprocess_features(mat_X_train, mat_X_test)
)

por_X_train_processed, por_X_test_processed, por_preprocessor = (
    preprocess_features(por_X_train, por_X_test)
)

print("Mathematics processed:", mat_X_train_processed.shape)
print("Portuguese processed:", por_X_train_processed.shape)

assert mat_X_train_processed.select_dtypes(
    exclude="number"
).empty

assert por_X_train_processed.select_dtypes(
    exclude="number"
).empty

print("Encoding and scaling verified.")
```

```python
def balance_training_data(X_train, y_train):
    smote = SMOTE(
        random_state=42,
        k_neighbors=5,
    )

    X_balanced, y_balanced = smote.fit_resample(
        X_train,
        y_train,
    )

    X_balanced = pd.DataFrame(
        X_balanced,
        columns=X_train.columns,
    )

    y_balanced = pd.Series(
        y_balanced,
        name="performance",
    )

    return X_balanced, y_balanced


mat_X_balanced, mat_y_balanced = balance_training_data(
    mat_X_train_processed,
    mat_y_train,
)

por_X_balanced, por_y_balanced = balance_training_data(
    por_X_train_processed,
    por_y_train,
)

print("Mathematics balanced:", mat_X_balanced.shape)
print(mat_y_balanced.value_counts().sort_index())

print("\nPortuguese balanced:", por_X_balanced.shape)
print(por_y_balanced.value_counts().sort_index())
```

```python
assert mat_X_balanced.shape[0] == len(mat_y_balanced)
assert por_X_balanced.shape[0] == len(por_y_balanced)

assert mat_y_balanced.value_counts().nunique() == 1
assert por_y_balanced.value_counts().nunique() == 1

assert len(mat_X_test_processed) == len(mat_y_test)
assert len(por_X_test_processed) == len(por_y_test)

MODEL_READY_DIR = PROCESSED_DIR / "model_ready"
MODEL_READY_DIR.mkdir(parents=True, exist_ok=True)

mat_X_balanced.to_csv(
    MODEL_READY_DIR / "mat_X_train_balanced.csv",
    index=False,
)
mat_y_balanced.to_csv(
    MODEL_READY_DIR / "mat_y_train_balanced.csv",
    index=False,
)

por_X_balanced.to_csv(
    MODEL_READY_DIR / "por_X_train_balanced.csv",
    index=False,
)
por_y_balanced.to_csv(
    MODEL_READY_DIR / "por_y_train_balanced.csv",
    index=False,
)

mat_X_test_processed.to_csv(
    MODEL_READY_DIR / "mat_X_test_processed.csv",
    index=False,
)
mat_y_test.to_frame().to_csv(
    MODEL_READY_DIR / "mat_y_test.csv",
    index=False,
)

por_X_test_processed.to_csv(
    MODEL_READY_DIR / "por_X_test_processed.csv",
    index=False,
)
por_y_test.to_frame().to_csv(
    MODEL_READY_DIR / "por_y_test.csv",
    index=False,
)

print("SMOTE verification successful.")
print("Balanced training data and untouched test data saved.")
```
