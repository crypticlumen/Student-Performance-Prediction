# Phase 5: Target Variable Creation

## Objective
Convert the final grade (G3) into a binary classification target.

- Pass: G3 >= 10
- At Risk: G3 < 10

```python
import pandas as pd
from pathlib import Path

# Locate the project directory
PROJECT_ROOT = Path.cwd().parent

# Load cleaned datasets
mat_path = PROJECT_ROOT / "data" / "processed" / "student_mat_clean.csv"
por_path = PROJECT_ROOT / "data" / "processed" / "student_por_clean.csv"

mat_df = pd.read_csv(mat_path)
por_df = pd.read_csv(por_path)

# Create binary target: 1 = Pass, 0 = At Risk
mat_df["performance"] = (mat_df["G3"] >= 10).astype(int)
por_df["performance"] = (por_df["G3"] >= 10).astype(int)

# Display class distributions
print("Mathematics dataset:")
print(mat_df["performance"].value_counts().sort_index())

print("\nPortuguese dataset:")
print(por_df["performance"].value_counts().sort_index())

# Verify the mapping
display(mat_df[["G3", "performance"]].head(10))
```

```python
# Save datasets with the newly created performance target

mat_output = PROJECT_ROOT / "data" / "processed" / "student_mat_target.csv"
por_output = PROJECT_ROOT / "data" / "processed" / "student_por_target.csv"

mat_df.to_csv(mat_output, index=False)
por_df.to_csv(por_output, index=False)

print("Target datasets saved successfully!")
print(f"Mathematics: {mat_output}")
print(f"Portuguese: {por_output}")
```

```python
# Reload saved files to verify persistence

mat_check = pd.read_csv(mat_output)
por_check = pd.read_csv(por_output)

assert "performance" in mat_check.columns
assert "performance" in por_check.columns

assert len(mat_check) == 395
assert len(por_check) == 649

print("Verification successful!")
print("Math shape:", mat_check.shape)
print("Portuguese shape:", por_check.shape)
print("Math missing targets:", mat_check["performance"].isna().sum())
print("Portuguese missing targets:", por_check["performance"].isna().sum())
```
